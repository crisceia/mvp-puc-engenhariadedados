# Carga de Dados
---
A carga inicial de dados foi realizada pela interface "Catálogo", através da criação de tabelas anuais na camada _bronze_, correspondente aos arquivos CSV disponibilizados no conjunto de dados Banco de Preços em Saúde (BPS), no [Portal de Dados Abertos do SUS](https://dadosabertos.saude.gov.br/dataset/bps).  

**Período de referência considerado**: 2024 a 2026.  

Dentro do "Catálogo", na aba "Detalhes", foi possível observar que a quantidade de registros total em cada uma das tabelas na camada _bronze_ corresponde ao número de registros nos arquivos originais em formato CSV.  

Além disso, também observamos que a estrutura e o tipo dos dados são os mesmos, mudando apenas o ano de referência.

Logo, criamos uma tabela unificada para o período de referência, na camada _silver_, conforme a seguir.  

```sql
CREATE OR REPLACE TABLE mvp_puc_engenhariadedados.silver.bps_2024_a_2026 AS

SELECT *, 2024 AS ano_arquivo FROM mvp_puc_engenhariadedados.bronze.bps_2024
UNION ALL
SELECT *, 2025 AS ano_arquivo FROM mvp_puc_engenhariadedados.bronze.bps_2025
UNION ALL
SELECT *, 2026 AS ano_arquivo FROM mvp_puc_engenhariadedados.bronze.bps_2026;
```

```sql
SELECT COUNT(*) AS total_registros
FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026;
```

Abaixo, validamos a unificação das tabelas anuais da camada _bronze_ para a tabela BPS na camada _silver_, no período de referência.  

É possível observar que o total de registros em cada uma das tabelas anuais corresponde à quantidade de registros da tabela unificada, estratificada por ano do arquivo (campo de controle criado no momento da unificação).

```sql
SELECT 
    2024 AS ano, 
    (SELECT COUNT(*) FROM mvp_puc_engenhariadedados.bronze.bps_2024) AS total_registros_original,
    (SELECT COUNT(*) FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026 WHERE ano_arquivo = 2024) AS total_registros_unificada
UNION ALL
SELECT 
    2025, 
    (SELECT COUNT(*) FROM mvp_puc_engenhariadedados.bronze.bps_2025),
    (SELECT COUNT(*) FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026 WHERE ano_arquivo = 2025)
UNION ALL
SELECT 
    2026, 
    (SELECT COUNT(*) FROM mvp_puc_engenhariadedados.bronze.bps_2026),
    (SELECT COUNT(*) FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026 WHERE ano_arquivo = 2026)
```
