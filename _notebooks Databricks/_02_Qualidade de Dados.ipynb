### 2.9 Validação do CNPJ da instituição, do fornecedor e do fabricante  

**NOTA**: primeiro é criada uma função para validação do CNPJ, considerando a quantidade de dígitos, a rejeição de sequências formadas por um único dígito e a validação dos dois dígitos verificadores. Após, a função foi aplicada separadamente aos CNPJs das instituições, fornecedores e fabricantes.  

```sql
CREATE OR REPLACE FUNCTION fn_valida_cnpj(cnpj_input STRING)
RETURNS BOOLEAN
LANGUAGE PYTHON
AS $$
import re

if cnpj_input is None:
    return None

# Remove caracteres que não sejam números
cnpj = re.sub(r'[^0-9]', '', cnpj_input)

# CNPJ deve possuir 14 dígitos
if len(cnpj) != 14:
    return False

# Rejeita sequências com todos os dígitos iguais
if len(set(cnpj)) == 1:
    return False

# Primeiro dígito verificador
pesos1 = [5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2]
soma = sum(int(cnpj[i]) * pesos1[i] for i in range(12))
resto = soma % 11
digito1 = 0 if resto < 2 else 11 - resto

# Compara com o primeiro dígito verificador informado
if digito1 != int(cnpj[12]):
    return False

# Segundo dígito verificador
pesos2 = [6, 5, 4, 3, 2, 9, 8, 7, 6, 5, 4, 3, 2]
soma = sum(int(cnpj[i]) * pesos2[i] for i in range(13))
resto = soma % 11
digito2 = 0 if resto < 2 else 11 - resto

# Compara com o segundo dígito verificador informado
if digito2 != int(cnpj[13]):
    return False

return True
$$;
```

```sql
SELECT cnpj_instituicao, COUNT(*) AS qtd_invalidos
FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026
WHERE cnpj_instituicao IS NOT NULL
  AND NOT fn_valida_cnpj(cnpj_instituicao)
GROUP BY cnpj_instituicao
ORDER BY qtd_invalidos DESC;
```

```sql
SELECT cnpj_fornecedor, COUNT(*) AS qtd_invalidos
FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026
WHERE cnpj_fornecedor IS NOT NULL
  AND NOT fn_valida_cnpj(cnpj_fornecedor)
GROUP BY cnpj_fornecedor
ORDER BY qtd_invalidos DESC;
```

```sql
SELECT cnpj_fabricante, COUNT(*) AS qtd_invalidos
FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026
WHERE cnpj_fabricante IS NOT NULL
  AND NOT fn_valida_cnpj(cnpj_fabricante)
GROUP BY cnpj_fabricante
ORDER BY qtd_invalidos DESC;
```

## 5. Outliers  

### 5.1 Identificação de itens com preço unitário potencialmente atípico

```sql
WITH registros_elegiveis AS (
    SELECT
        co_seq_bps,
        co_catmat,
        CAST(vl_preco_unitario AS DECIMAL(18,6))
    FROM mvp_puc_engenhariadedados.silver.bps_2024_a_2026
    WHERE co_catmat IS NOT NULL
      AND vl_preco_unitario IS NOT NULL
      AND vl_preco_unitario > 0
),

catmats_elegiveis AS (
    SELECT
        co_catmat
    FROM registros_elegiveis
    GROUP BY co_catmat
    HAVING COUNT(*) >= 5
),

dados_elegiveis AS (
    SELECT
        r.co_seq_bps,
        r.co_catmat,
        r.vl_preco_unitario
    FROM registros_elegiveis r
    INNER JOIN catmats_elegiveis c
        ON r.co_catmat = c.co_catmat
),

estatisticas AS (
    SELECT
        co_catmat,
        PERCENTILE_APPROX(vl_preco_unitario, 0.25, 10000) AS q1,
        PERCENTILE_APPROX(vl_preco_unitario, 0.75, 10000) AS q3
    FROM dados_elegiveis
    GROUP BY co_catmat
),

classificacao AS (
    SELECT
        d.co_seq_bps,
        d.co_catmat,
        d.vl_preco_unitario,
        CASE
            WHEN d.vl_preco_unitario < e.q1 - 1.5 * (e.q3 - e.q1)
                THEN 'Abaixo do limite inferior'
            WHEN d.vl_preco_unitario > e.q3 + 1.5 * (e.q3 - e.q1)
                THEN 'Acima do limite superior'
        END AS tipo_atipicidade
    FROM dados_elegiveis d
    INNER JOIN estatisticas e
        ON d.co_catmat = e.co_catmat
)

SELECT
    COUNT(*) FILTER(
        WHERE tipo_atipicidade = 'Abaixo do limite inferior'
    ) AS qtd_potenciais_outliers_inferiores,

    COUNT(*) FILTER(
        WHERE tipo_atipicidade = 'Acima do limite superior'
    ) AS qtd_potenciais_outliers_superiores
FROM classificacao;
```
